```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
import math

np.random.seed(42)  # remove or change for different samples
data = np.load("bc_dataset_clean.npz", allow_pickle=False)
frames = data["frames"]

# random unique indices, then sort
idxs = np.random.choice(len(frames), size=15, replace=False)
idxs.sort()  # ensures order like 5, 100, 1000 ...

root = Path("")  # set if paths are relative

cols = 4
rows = math.ceil(len(idxs) / cols)
fig, axes = plt.subplots(rows, cols, figsize=(cols*3, rows*3))
axes = axes.ravel()

for i, ax in enumerate(axes):
    ax.axis("off")
    if i >= len(idxs): 
        continue
    idx = int(idxs[i])
    p = root / Path(str(frames[idx]))
    try:
        img = Image.open(p).convert("RGB")
        ax.imshow(img)
        ax.set_title(f"{idx}", fontsize=8)
    except Exception as e:
        ax.text(0.5, 0.5, f"Err:\n{e}", ha="center", va="center", fontsize=8)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
from pathlib import Path
import math

data = np.load("bc_dataset_clean.npz", allow_pickle=False)
frames = data["frames"]
root = Path("")

rng = np.random.default_rng()
idxs = np.sort(rng.choice(len(frames), size=15, replace=False))

cols = 4
rows = math.ceil(len(idxs) / cols)
fig, axes = plt.subplots(rows, cols, figsize=(cols*3, rows*3))
axes = axes.ravel()

for i, ax in enumerate(axes):
    ax.axis("off")
    if i >= len(idxs):
        continue
    idx = int(idxs[i])
    p = root / Path(str(frames[idx]))
    try:
        img = Image.open(p).convert("RGB")
        img = ImageOps.exif_transpose(img)   # correct EXIF orientation first
        img = img.transpose(Image.ROTATE_90) # rotate 90° anticlockwise
        ax.imshow(img)
        ax.set_title(f"{idx}", fontsize=8)
    except Exception as e:
        ax.text(0.5, 0.5, f"Err:\n{e}", ha="center", va="center", fontsize=8)

plt.tight_layout()
plt.show()
```
